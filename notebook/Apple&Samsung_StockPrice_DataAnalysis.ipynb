# **Risk-Adjusted Return (Sharpe Ratio)**

A higher average return means little without knowing how much risk was taken to get it. The Sharpe Ratio expresses return per unit of volatility (risk-free rate assumed to be 0% for simplicity), making it easier to judge which stock rewarded investors more efficiently for the risk involved.


```python
apple_sharpe = apple['Change %'].mean() / apple['Change %'].std()
samsung_sharpe = samsung['Change %'].mean() / samsung['Change %'].std()

print(f"Apple Sharpe Ratio (monthly): {apple_sharpe:.2f}")
print(f"Samsung Sharpe Ratio (monthly): {samsung_sharpe:.2f}")
```

# **Compound Annual Growth Rate (CAGR)**

```python
years = (apple['Date'].iloc[-1] - apple['Date'].iloc[0]).days / 365.25

apple_cagr = (
    (apple['Price'].iloc[-1] / apple['Price'].iloc[0]) ** (1/years) - 1
) * 100

samsung_cagr = (
    (samsung['Price'].iloc[-1] / samsung['Price'].iloc[0]) ** (1/years) - 1
) * 100

print(f"Period covered: {years:.2f} years")
print(f"Apple CAGR: {apple_cagr:.2f}%")
print(f"Samsung CAGR: {samsung_cagr:.2f}%")
```

# **Maximum Drawdown**

```python
def max_drawdown(price):

    running_max = price.cummax()

    drawdown = (price - running_max) / running_max

    return drawdown.min()*100

apple_dd = max_drawdown(apple['Price'])

samsung_dd = max_drawdown(samsung['Price'])

print(f"Apple Maximum Drawdown: {apple_dd:.2f}%")
print(f"Samsung Maximum Drawdown: {samsung_dd:.2f}%")
```

# **Statistical Test**

```python
from scipy.stats import ttest_ind

t_stat,p_value = ttest_ind(
    apple['Change %'],
    samsung['Change %']
)

print("T Statistic :",t_stat)

print("P Value :",p_value)
```

**Interpreting the test:** The independent two-sample t-test compares the average monthly returns of Apple and Samsung. If the p-value is below 0.05, the difference in mean monthly returns between the two stocks is statistically significant at the 95% confidence level, meaning it is unlikely to be due to random chance. If the p-value is above 0.05, we cannot conclude the two stocks have significantly different average returns based on this sample — any observed difference could plausibly be noise.


# **Key Findings**

The analysis of Apple's and Samsung's historical stock data produced several important findings regarding their market performance, investment potential, and risk characteristics.

- Apple and Samsung delivered **nearly identical annualized growth** over the period (Apple CAGR: 14.48%, Samsung CAGR: 14.38%), so long-term growth alone does not meaningfully separate the two stocks.
- The two companies differ far more on **risk than on return**. Apple's monthly returns were more stable, while Samsung showed larger positive and negative monthly movements.
- Moving average analysis confirmed that Apple maintained a smoother upward trend, whereas Samsung exhibited more frequent short-term reversals.
- Volatility analysis and the Sharpe Ratio (Apple: 0.19, Samsung: 0.16) both indicate that Apple delivered a similar return per unit of risk taken — slightly more efficiently than Samsung.
- Trading volume increased during periods of significant price movement for both companies, suggesting higher investor activity during volatile market conditions.
- The correlation between Apple's and Samsung's monthly returns was **0.25 — a weak-to-moderate positive relationship**, meaning the two stocks move somewhat together but far from in lockstep, which supports a modest diversification benefit from holding both.
- Maximum Drawdown analysis showed Samsung's largest peak-to-trough decline (-37.54%) was substantially deeper than Apple's (-26.83%), highlighting Samsung's greater downside risk.
- A t-test comparing average monthly returns produced a p-value of 0.90, meaning the difference in average returns between the two stocks is **not statistically significant** — any apparent gap in average return is plausibly due to chance, not a real underlying difference.

### Overall Finding

Overall, the results suggest that **Apple and Samsung delivered comparable long-term growth, but Apple did so with meaningfully lower risk** — a smaller maximum drawdown and a slightly better Sharpe Ratio. Samsung is not the "higher growth" option here; it is the higher-volatility option for a similar level of return. Both companies remain major players in the global technology sector, and their modest (0.25) return correlation means a diversified position in both may still help investors manage portfolio-level risk.
