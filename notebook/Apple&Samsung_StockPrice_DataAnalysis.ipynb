```python
plt.style.use('seaborn-v0_8-whitegrid')

apple["MA6"]=apple["Price"].rolling(6).mean()
apple["MA12"]=apple["Price"].rolling(12).mean()

plt.figure(figsize=(14,6))

plt.plot(apple["Date"],apple["Price"],label="Price")

plt.plot(apple["Date"],apple["MA6"],label="6 Month MA")

plt.plot(apple["Date"],apple["MA12"],label="12 Month MA")

plt.legend()

plt.grid(False)

plt.title("Apple Moving Average")

plt.show()




plt.style.use('seaborn-v0_8-whitegrid')

samsung["MA6"]=samsung["Price"].rolling(6).mean()
samsung["MA12"]=samsung["Price"].rolling(12).mean()

plt.figure(figsize=(14,6))

plt.plot(samsung["Date"],samsung["Price"],label="Price")

plt.plot(samsung["Date"],samsung["MA6"],label="6 Month MA")

plt.plot(samsung["Date"],samsung["MA12"],label="12 Month MA")

plt.legend()

plt.grid(False)

plt.title("Samsung Moving Average")

plt.show()
```

# **Risk vs Return Analysis**

```python
risk_return = pd.DataFrame({
    'Company':['Apple','Samsung'],
    'Average Return (%)':[
        apple['Change %'].mean()*100,
        samsung['Change %'].mean()*100
    ],
    'Risk (Std Dev %)':[
        apple['Change %'].std()*100,
        samsung['Change %'].std()*100
    ]
})

print(risk_return)




```

```python
plt.style.use('seaborn-v0_8-whitegrid')

plt.figure(figsize=(7,6))


point_colors = ['lightcoral', 'skyblue']

for i in range(len(risk_return)):
    plt.scatter(
        risk_return['Risk (Std Dev %)'][i],
        risk_return['Average Return (%)'][i],
        s=250,
        color=point_colors[i]
    )
    plt.text(
        risk_return['Risk (Std Dev %)'][i],
        risk_return['Average Return (%)'][i],
        risk_return['Company'][i],
        fontsize=11
    )

plt.xlabel('Risk (Standard Deviation %)')
plt.ylabel('Average Monthly Return (%)')
plt.title('Risk vs Return Analysis')
plt.grid(False)

plt.show()
```

# **Compound Annual Growth Rate (CAGR)**

```python
years = 5

apple_cagr = (
    (apple['Price'].iloc[-1] / apple['Price'].iloc[0]) ** (1/years) - 1
) * 100

samsung_cagr = (
    (samsung['Price'].iloc[-1] / samsung['Price'].iloc[0]) ** (1/years) - 1
) * 100

print(f"Apple CAGR: {apple_cagr:.2f}%")
print(f"Samsung CAGR: {samsung_cagr:.2f}%")
```

# **Maximum Drawdown**

```python
def max_drawdown(price):

    running_max = price.cummax()

    drawdown = (price - running_max) / running_max

    return drawdown.min()*100

apple_dd = max_drawdown(apple['Price'])

samsung_dd = max_drawdown(samsung['Price'])

print(f"Apple Maximum Drawdown: {apple_dd:.2f}%")
print(f"Samsung Maximum Drawdown: {samsung_dd:.2f}%")
```
